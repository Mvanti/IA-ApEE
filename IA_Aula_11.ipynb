## Introdução ao Pandas DataFrame

O Pandas é uma biblioteca Python para trabalhar com dados organizados em tabelas. Considere, por exemplo, a seguinte tabela:

| Nome  | Idade | Curso      | Nota |
| ----- | ----: | ---------- | ---: |
| Ana   |    20 | Engenharia |  8.5 |
| João  |    22 | Computação |  7.0 |
| Maria |    21 | Engenharia |  9.0 |
| Pedro |    23 | Computação |  6.5 |

Este tabela pode ser representada por um DataFrame, ou uma estrutura de dados tabelado no Python. 



```python
import pandas as pd
```

#### Criação de um DataFrame a partir de um dicionário

```python
import pandas as pd

dados = {
    'Nome': ['Ana', 'João', 'Maria', 'Pedro'],
    'Idade': [20, 22, 21, 23],
    'Curso': ['Engenharia Elétrica', 'Computação',
              'Engenharia de Telecomunicações', 'Computação'],
    'Nota': [8.5, 7.0, 9.0, 6.5]
}
print(dados)
```

O DataFrame é construído a partir do dicionário

```python
df=pd.DataFrame(dados)
print(df)
## Importante: os índices 0-3 identificando as linhas são criados automaticamente pelos Pandas
```

Acessando uma coluna

```python
df['Nome']
```

```python
notas=df['Nota']
print(notas)
```

Seleção de várias colunas

```python
df[['Nome', 'Idade', 'Nota']] # atenção ao duplo colchete quando várias colunas são selecionadas!
```

Seleção de linhas

```python
df.loc[0]
```

Seleção de várias linhas 

```python
df.loc[0:2]
```

```python
df.iloc[0:2]
```

### Importante:

| Sintaxe                 | Índices retornados |
| ----------------------- | ------------------ |
| `a[0:2]` — NumPy        | 0, 1               |
| `df.iloc[0:2]` — Pandas | 0, 1               |
| `df.loc[0:2]` — Pandas  | **0, 1, 2**        |


Filtragem de dados

Exemplo: Recuperação dos alunos com nota > 7:

```python
df[df['Nota'] > 7]
```

Note que o resultado da expressão no par de colchetes interno resulta em:
True
False
True
False

#### Criação de um DataFRame a partir de um *array* 2D

```python
import numpy as np

data = np.array([[1, 4], [2, 5], [3, 6]])

df=pd.DataFrame(data)
df
```

Formatação do DataFrame

```python
df = pd.DataFrame(data, index=['Linha1', 'Linha2', 'Linha3'],columns=['Col1', 'Col2'])
df
```
