## O Perceptron 
O perceptron é um dos primeiros modelos computacionais do que é hoje conhecido como uma rede neural. Os fundamentos de operação do perceptron são baseados na estrutura anatômica de um único nerônio.

O neurônio, ou célula neuronal, é uma célula excitável pela soma dos sinais elétricos recebidos de neurônios vizinhos através de fibras denominadas de dentrites, como pode ser visto na figura acima. Basicamente, quando a soma dos sinais atinge um determinado limiar, o neurônio "dispara" um impulso elétrico através do Axônio, a fibra mais longa mostrada na figura. 

O Perceptron foi proposto por Frank Rosemblat em 1957, concebido como um dispositivo eletrônico construído de acordo com os princípios biológicos de neurônio, tendo mostrado habilidades para aprender. Este dispositivo foi então simulado em um IBM 704. 
Como seu modelo biológico, o modelo recebe *m* entradas e libera um valor de saída. Mais especificamente, o perceptron faz a soma ponderada dos dados de entrada multipicando entradas $x_i$ por um peso correspondente $w_i$,
$$z=\sum_{i=1}^m w_ix_i$$

Quando esta soma excede um limiar pre-determinado $\theta$, um valor é entregue na saída. Este comportamento pode ser emulado com a aplicação de uma função de ativação $f_a$ de forma que:
$$ \hat{y}=\begin{cases} 0 & z<\theta\\f_a(\sum_{i=1}^m w_ix_i),& z \ge \theta\end{cases}$$

Em problemas de classificação binária utiliza-se a função degrau ou função de Heaviside como função de ativação, de forma que 
$$\hat{y}=\begin{cases}0,&z<\theta\\1, &z\ge \theta\end{cases}$$
Desta forma, a saída $\hat{y}$ apontará para uma das duas classes do problema. 

O objetivo do modelo é implementar um algoritmo para determinação dos pesos, os quais com um limiar apropriado seja capaz de reproduzir as saídas corretas a partir de entradas conhecidas. Este algoritmo executa então o processo de treinamento, e para simplifica-lo, adicionamos outro peso, denominado *bias*, representado pela letra *b,* conectado á uma nova entrada de valor unitário. Ou seja, 
$$ z=\sum_{i=1}^m w_ix_i+b=\sum_{i=0}^m w_ix_i$$

Desta forma, o *bias* é treinado como os outros pesos, e podemos assim utilizar $\theta=0 $. 
Durante o treinamento, para cada amostra $(X_i,y_i)$, o vetor de pesos $\textbf{w}=w_0,w_1,\textellipsis$ é atualizado segundo a expressão simples abaixo:
$$\textbf{w}\leftarrow\textbf{w}+ \alpha(y_i-\hat{y}_i)X_i$$
onde $\alpha$ é a taxa de aprendizado que controla o comprimento do ajuste do peso para cada atualização. O teorema de convergência do perceptron mostra que em um problema de classificação, quando os dados são linearmente separáveis,  o algoritmo irá eventualmente convergir para os pesos de um hiperplano separando os dados segundo suas categorias, em um número finito de iterações (https://www.cs.columbia.edu/~mcollins/courses/6998-2012/notes/perc.converge.pdf).

Considere a classe Perceptron abaixo:

```python
import numpy as np

class Perceptron:
    # inicialização dos atributos da classe
    def __init__(self, learning_rate=0.01, epochs=1000, activ=True):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None
        self.activ=activ
        
    #Função de treinamento
    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0
        # Para cada época. itera sobre todos as amostras de entrada, calcula o erro e atualiza os pesos
        for _ in range(self.epochs):
            for idx, x_i in enumerate(X):
                if self.activ:
                    # Classificação binária
                    y_predicted = self.step_function(np.dot(x_i, self.weights) + self.bias)
                else:
                    # Regressão  Linear
                    y_predicted = np.dot(x_i, self.weights) + self.bias
    
                # Atualiza os pesos
                update = self.learning_rate * (y[idx] - y_predicted)
                self.weights += update * x_i
                self.bias += update

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        if self.activ:
            y_predicted = self.step_function(linear_output)
        else:
            y_predict=linear_output
        return y_predicted
    def step_function(self, x):
        return np.where(x >= 0, 1, 0)
```

Veja a função `__init__ `na classe acima. As definições self.variável fazem com que variável seja um atributo da classe. Assim, se fizermos           `p = Perceptron(learning_rate=0.1, epochs=100)`, p é um objeto da classe perceptron com os atributos declarados, acessíveis como p.learning_rate, p.epochs, etc. Em aprendizado profundo, as **épocas** são os laços iterativos em cada qual todos os pesos são atualizados de acordo com todos os dados e rótulos de entrada. 
Os exemplos a seguir ilustrarão a aplicação do perceptron no aprendizado da operação de portas lógicas. Considere incialmente a porta **AND**, com as regras definidas em X e y abaixo:

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])
p = Perceptron(learning_rate=0.1, epochs=10)
p.fit(X,y)
print("Predictions:", p.predict(X))
```

### O Perceptron aplicado à um problema de regressão linear

Para verificar a aolicação do perceptron em um problema deregressão linear, repetiremos aqui o problema de nossa aula sobre regressão linear no qual utilizamos a equação normal para derivar $\beta_0 $ e $\beta_1$ para os dados `X=np.array([[1],[2],[3]])` e `y=np.array([[2],[3],[5]])´. Com o perceptron, teremos $w_0=b=\beta_0\;\text{e}\;w_1=\beta_1$. Observe também que a função de ativação degrau resultaria apenas nas saídas 1 e 0, quando agora a saída deve ser uma função contínua. Por isso substituímos a função de ativação degrau pela função linear, ou seja, $f_a(z)=z$.

```python
X=np.array([[1],[2],[3]])
y=np.array([[2],[3],[5]])
p = Perceptron(learning_rate=0.1, epochs=10, activ=False)
p.fit(X,y)
print("b=",p.bias)
print("w=",p.weights)
```

Em uma rede neural o treinamento deve ser calculado com o gradiente do erro, como será visto posteriormente com o método dos gradientes descendentes.
