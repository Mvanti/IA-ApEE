## O Perceptron Multicamada

É um tipo de rede neural consistindo de múltiplas camadas de neurônios. Nesta rede, os dados presentes nos neurônios na camada de entrada são ponderados e somados em cada neurônio da primeira camada intermediária, avaliados por uma função de ativação $f_a$ neste neurônio e então encaminhado para as camadas seguintes. Este processo se repete até os neurônios da camada de saída, resultando em um valor de $\hat{y}_i$ em dada um. Este processo é denotado de *feedforward*. 
No treinamento, os gradientes da função de erro ou perda emrelação aos pesos são calculados durante a retropropagação, ou *backpropagation*.

Nesta rede temos dois neurônios na camada inicial, três na camada intermediária de um na da saída. Os pesos conectam os  neurônios de cada camada a todos os neurônios na camada seguinte. Em cada conexão entre o  i-ésimo neurônio da camada $k+1$ e o j-ésimo neurônio da camada $k$ o sinal é ponderado pelo peso $W_{ij}^k$. Desta forma, o primeiro neurônio da camada inicial na rede acima é conectado a camada seguinte através dos pesos $W_{11}^1$, $W_{21}^1$ e $W_{31}^1$. Para o segundo neurônio da camada de entrada teremos $W_{12}^1$, $W_{22}^1$ e $W_{32}^1$. Além disso, em cada neurônio na camada intermediária há um *bias* $b_i^1$. 

### *Feedforward*

Na rede acima há uma camada intermediária. Em cada neurônio desta camada é acumulada a soma ponderada de suas entradas. Se as entradas da rede são $X_1$ e $X_2$, então teremos para o i-ésimo neurônio da camada intermediária
$$Z_i=W_{i1}^1X_1+ W_{i2}^1X_2+b_i$$

O i-ésimo neurônio aplica uma função de ativação $f_z$ à  $Z_i$, resultando em $h_i$, a saída deste neurônio da camada intermediária, ou seja , 
$$h_i=f_a(Z_i).$$
