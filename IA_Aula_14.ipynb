## O Perceptron Multicamada

É um tipo de rede neural consistindo de múltiplas camadas de neurônios. Nesta rede, os dados presentes nos neurônios na camada de entrada são ponderados e somados em cada neurônio da primeira camada intermediária, avaliados por uma função de ativação $f_a$ deste neurônio e então encaminhado para as camadas seguintes. Este processo se repete até os neurônios da camada de saída, resultando em um valor de $\hat{y}_i$ em cada um. Este processo é denotado de *feedforward*. 
No treinamento, os gradientes da função de erro ou perda em relação aos pesos são calculados durante a retropropagação, ou *backpropagation*.

Nesta rede temos dois neurônios na camada inicial, três na camada intermediária e um na da saída. Os neurônios de cada camada são conectados a todos os neurônios na camada seguinte. Em cada conexão entre o  i-ésimo neurônio da camada $k+1$ e o j-ésimo neurônio da camada $k$ o sinal é ponderado pelo peso $W_{ij}^k$. Desta forma, o primeiro neurônio da camada inicial na rede acima é conectado a camada seguinte através dos pesos $W_{11}^1$, $W_{21}^1$ e $W_{31}^1$. Para o segundo neurônio da camada de entrada teremos $W_{12}^1$, $W_{22}^1$ e $W_{32}^1$. Além disso, em cada neurônio na camada intermediária há um *bias* $b_i^1$. 

### *Feedforward*

Na rede acima há uma camada intermediária. Em cada neurônio desta camada é acumulada a soma ponderada de suas entradas. Se as entradas da rede são $X_1$ e $X_2$, então teremos para o i-ésimo neurônio da camada intermediária
$$z_i=W_{i1}^1X_1+ W_{i2}^1X_2+b_i$$

O i-ésimo neurônio aplica uma função de ativação $f_a$ à  $z_i$, resultando em $h_i$, a saída deste neurônio da camada intermediária, ou seja , 
$$h_i=f_a(z_i).$$

 #### Algumas funções de ativação mais frequentemente usadas

 * Função Linear:
   $$ f_a(z)=z$$
   
   A função linear ou identidade apenas repete o valor de $z$ em cada neurônio onde ela é aplicada. Como não introduz não linearidades, ela não é capaz de tratar problemas não lineares, e camadas internas empilhadas com função linear não melhora o resultado. Uma vantagem deste tipo de função é apresentar derivada constante, podendo assim evitar gradientes que explodem ou desaparecem.

* Sigmoide

 $$f_a(z)=\frac{1}{1+e^{-z}}=\frac{e^z}{1+e^{z}}$$
 Esta é uma função não linear sendo por isso útil em problemas que demandam uma resposta não linear. Além disso, como $0 \le f_a(z)\le 1$ esta função pode ser interpretada como uma distribuição de probabilidade, o que é  muito útil em problemas de classificação.

 Os gradientes da função sigmoide podem ser reduzidos desacelerando consequentemente a convergência no treinamento. Além disso seu uso na camada de saída pode ser ineficiente devido a sua pequena faixa de valores. 

* Tangente Hiperbólica
  $$ f_a(z)=\tanh{z}=\frac{e^{z}-e^{-z}}{e^z+e^{-z}}$$

  Muitas vezes preferida em relação à sigmoide quando os dados de entrada são centrados, também apresenta maiores derivadas que a sigmoide, mas ainda sendo susceptível aos *vanishing gradients*.
* Relu (*Rectified Linear Unit* )
  $$f_a(z)=\begin{cases}0,&z\le0\\z,&z>0\end{cases}$$
  É uma função não linear esparsa, ativando apenas neurônios com entradas positivas. Muito eficiente nas camadas internas, tende a evitar os *vanishing gradients*. 
 ##### **Exemplo** 
 Considere a rede acima com todos os pesos da camada 1 unitários, $b_1^1=b_3^1=b_1^2=1$, $b_2^1=-1$, $W_{12}^2=-2$, $W_{11}^2=W_{13}^2=1$, $X_1=1$ e $X_2=0$, e com funções de ativação relu na camada intermediária e Heaviside na camada de saída. Faça o *feedforward* e determine a saída $\hat{y}$.

* Na camada interna escondida, 
 $$h_1=f_a(z_1)=f_a(W_{11}^1X_1+ W_{12}^1X_2+b_1^1)=f_a(X_1+b_1^1)=f_a(1+1)=2$$
 $$h_2=f_a(z_2)=f_a(W_{21}^1X_1+ W_{22}^1X_2+b_2^1)=f_a(X_1+b_2^1)=f_a(1-1)=0$$
 $$h_3=f_a(z_3)=f_a(W_{31}^1X_1+ W_{32}^1X_2+b_3^1)=f_a(X_1+b_3^1)=f_a(1+1)=2$$
* Na camada de saída,
$$\hat{y}=f_a(z_1^2)=f_a(W_{11}^2h_1+ W_{12}^2h_2+W_{13}^2h_3+b_1^2)=f_a(h_1-2h_2 +h_3 +b_1^2)=f_a(2-0+2+1)=1$$

### Retropropagação ou *Backpropagation* 

A retropropagação é o algoritmo no qual a rede neural é treinada. Durante a execução do algoritmo, os gradientes de uma função de perdas $L(\hat{y})$ é calculado em função das derivadas em relação à todos os pesos e *bias* da rede. 
Os gradientes são calculados por meio de derivadas parciais, inciando com os pesos da última camada e retornando retroativamente até os pesos da primeira camada, sendo por isso que o algoritmo leva este nome. O método dos gradientes descendentes é então utilizado para reduzir a função de perdas. 

Como exemplo iremos obter o gradientes da rede acima $\frac{\partial L}{\partial W_{11}^2}$ e $\frac{\partial L}{\partial W_{11}^1}$.
* Como primeiro passo, calculamos a função de perdas utilizando o erro médio quadrático *mse*, ou seja :
  $$L(\hat{y})=\frac{1}{N}\sum_{i=1}^N\|y_i-\hat{y}_i\|^2$$
* Camada 2: $\frac{\partial L}{\partial W_{11}^2}.\;$

  Para cada par (amostra) $X_1$ e $X_2$ calculamos:
  $$\frac{\partial L}{\partial W_{11}^2}=\frac{\partial L}{\partial \hat{y}_i}\frac{\partial \hat{y}_i}{\partial W_{11}^2}$$
  onde
  $$\frac{\partial L}{\partial \hat{y}_i}=-2(y_i-\hat{y}_i)$$
  e
  $$\frac{\partial \hat{y}_i}{\partial W_{11}^2}=\frac{\partial f_a(z_1^2)}{\partial z_1^2}\frac{\partial z_1^2}{\partial W_{11}^2}$$
  Ora, $z_1^2= W_{11}^2h_1+ W_{12}^2h_2+W_{13}^2h_3+b_1^2$, portanto
  $$\frac{\partial z_1^2}{\partial W_{11}^2}=h_1$$
Assim,
 $$\frac{\partial \hat{y}_i}{\partial W_{11}^2}=\frac{\partial f_a(z_1^2)}{\partial z_1^2}h_1$$

e 

$$ \frac{\partial L}{\partial W_{11}^2}=-2(y_i-\hat{y}_i)\frac{\partial f_a(z_1^2)}{\partial z_1^2}h_1$$

* Camada 1: $\frac{\partial L}{\partial W_{11}^1}.\;$
 
$$\frac{\partial L}{\partial W_{11}^1}=\frac{\partial L}{\partial \hat{y}_i}\frac{\partial \hat{y}_i}{\partial h_1}\frac{\partial h_1}{\partial W_{11}^1},$$
mas 
$$\frac{\partial L}{\partial \hat{y}_i}=-2(y_i-\hat{y}_i),$$
$$\frac{\partial \hat{y}_i}{\partial h_1}=\frac{\partial f_a(z_1^2)}{\partial z_1^2}\frac{\partial z_1^2}{\partial h_1}=\frac{\partial f_a(z_1^2)}{\partial z_1^2}W_{11}^2\;\text{e}$$
$$\frac{\partial h_1}{\partial W_{11}^1}=\frac{\partial f_a(z_1^1)}{\partial z_1^1}\frac{\partial z_1^1}{\partial W_{11}^1}=\frac{\partial f_a(z_1^1)}{\partial z_1^1}X_1$$
Juntando tudo, dá:
$$\frac{\partial L}{\partial W_{11}^1}=-2(y_i-\hat{y}_i)W_{11}^2\frac{\partial f_a(z_1^2)}{\partial z_1^2}\frac{\partial f_a(z_1^1)}{\partial z_1^1}X_1$$
Note na equação acima que as duas funções de ativação não são geralmente iguais, sendo qua sus derivadas também serão diferentes. Calculando essas derivadas, e expandindo este desenvolvimento para todos os pesos e *bias*, obtemos o vetor gradiente.
